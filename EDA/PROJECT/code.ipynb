```python
# checking null values if exist
df.isna().sum()
```

```python
# handling null values

for col in df.columns:
    mode_value = df[col].mode()
    if not mode_value.empty:
        df[col] = df[col].fillna(mode_value[0])

df.isna().sum()
```

```python
# unique

for col in df.columns:
    print(col)
    print(df[col].unique())
    print("_"*100)
```

```python
# outlier detection

num_col = df.select_dtypes(include=np.number).columns

for col in num_col:
  plt.figure()
  sns.boxplot(x = df[col])
  plt.title(f"Boxplot of {col}")
  plt.show()
```

```python
# removing outlier using IQR method
for c in num_col:
  q1 = df[c].quantile(0.25)
  q3 = df[c].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5 * iqr
  upper_bound = q3 + 1.5 * iqr
  df[c] = df[c].clip(lower_bound, upper_bound)
```

```python
# boxplot of removed outlier
for col in num_col:
  plt.figure()
  sns.boxplot(x = df[col])
  plt.title(f"Boxplot of {col}")
  plt.show()
```

```python
# checking correlation
num_df = df.select_dtypes(include="number")

# Find correlation
corr = num_df.corr()

corr
```

```python
# visualize correlation using heatmap

corr = df.corr(numeric_only=True)
plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap - strategic preparation in athletics")
plt.show()
```

```python
# scatter plot - Training hours vs performance score

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df.sample(min(3000, len(df)), random_state=42),
    x="Training_Hours_Per_Week",
    y="Performance_Score",
    alpha=0.5
)

plt.title("Training Hours vs Performance Score")
plt.xlabel("Training Hours Per Week")
plt.ylabel("Performance Score")
plt.tight_layout()
plt.show()
```

```python
# histogram - Distributions of performance scores

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Performance_Score",
    bins=30,
    kde=True
)

plt.title("Distribution of Performance Scores")
plt.xlabel("Performance Score")
plt.ylabel("Number of Athletes")
plt.tight_layout()
plt.show()
```

```python
# bar chart - Average performance score by sport type

plt.figure(figsize=(10, 5))

avg_performance = (
    df.groupby("Sport_Type")["Performance_Score"]
    .mean()
    .sort_values(ascending=False)
)

sns.barplot(
    x=avg_performance.index,
    y=avg_performance.values
)

plt.title("Average Performance Score by Sport Type")
plt.xlabel("Sport Type")
plt.ylabel("Average Performance Score")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# line chart - Training hours vs average performance trend

plt.figure(figsize=(9, 5))

temp = df[["Training_Hours_Per_Week", "Performance_Score"]].dropna().copy()

temp["Training_Hours_Group"] = (
    temp["Training_Hours_Per_Week"].round()
)

trend = (
    temp.groupby("Training_Hours_Group")["Performance_Score"]
    .mean()
    .sort_index()
)

plt.plot(
    trend.index,
    trend.values,
    marker="o",
    linewidth=2
)

plt.title("Training Hours vs Average Performance Trend")
plt.xlabel("Training Hours Per Week (Rounded)")
plt.ylabel("Average Performance Score")
plt.tight_layout()
plt.show()
```

```python
# pie chart - Athlete distribution by gender

plt.figure(figsize=(7, 7))

gender_counts = df["Gender"].dropna().value_counts()

plt.pie(
    gender_counts.values,
    labels=gender_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Athlete Distribution by Gender")
plt.axis("equal")
plt.show()
```
